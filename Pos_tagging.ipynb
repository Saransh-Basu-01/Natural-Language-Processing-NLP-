```python
import spacy
nlp=spacy.load('en_core_web_sm')
```

```python
doc=nlp(u"I will google about facebook")
```

```python
doc.text
```

```python
doc[0]
```

```python
doc[0].pos_
```

```python
doc[0].tag_
```

```python
spacy.explain('PRP') #fine grain part of speech
```

```python
for word in doc:
  print(word.text,"----->",word.pos_,word.tag_,spacy.explain(word.tag_))
```

```python
doc2=nlp("i left the room")
doc3=nlp("to the left of the room")
```

```python
for word in doc2:
  print(word.text,"----->",word.pos_,word.tag_,spacy.explain(word.tag_))
```

```python
for word in doc3:
  print(word.text,"----->",word.pos_,word.tag_,spacy.explain(word.tag_))
```
